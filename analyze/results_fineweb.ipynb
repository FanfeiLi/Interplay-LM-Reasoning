# FineWeb-Edu: Paradigm Lines on Downstream Benchmarks

Compare AR (Transformer, Mamba) vs Diffusion (MDLM, BD3LM) models
trained on FineWeb-Edu with the same Qwen2.5 tokenizer.

Plot benchmark-vs-benchmark to see if different paradigms lie on different lines.

```python
import os, sys
os.chdir("/fast/pmayilvahanan/Interplay-LM-Reasoning")
sys.path.insert(0, "analyze")

from results_fineweb import *
%matplotlib inline
```

```python
df = load_all()
if len(df) > 0:
    print(df.groupby(["model_type", "size"]).size())
else:
    print("No results yet. Run training and evaluation first.")
```

## Benchmark vs Benchmark Plots

```python
if len(df) > 0:
    pairs = [
        ("arc_easy", "arc_challenge"),
        ("hellaswag", "mmlu"),
        ("piqa", "winogrande"),
        ("hellaswag", "arc_easy"),
        ("arc_easy", "mmlu"),
        ("hellaswag", "piqa"),
    ]
    for task_x, task_y in pairs:
        fig = plot_benchmark_vs_benchmark(df, task_x, task_y)
        if fig is not None:
            plt.show()
```

## NLL vs Accuracy

```python
if len(df) > 0:
    for task in ["hellaswag", "arc_easy", "arc_challenge", "mmlu"]:
        fig = plot_loss_vs_accuracy(df, task)
        if fig is not None:
            plt.show()
```

## Summary Table

```python
if len(df) > 0:
    acc_cols = [c for c in df.columns if c.endswith("_acc")]
    summary = df.groupby(["family", "size"])[acc_cols].max()
    summary = (summary * 100).round(1)
    summary.columns = [c.replace("_acc", "") for c in summary.columns]
    display(summary)
```
